# Part 4: Test Design Practice (40 minutes)

## Exercise 4.1: Strategic Test Design with Risk Analysis

**Scenario:** You need to test a password strength validator.

**Requirements:**
- Password must be at least 8 characters
- Must contain at least one uppercase letter
- Must contain at least one number
- Must contain at least one special character (!@#$%^&*)

### Step 1: Risk Assessment

````python
"""
Risk Analysis:

1. What happens if this function fails?
   - Security implications: if it wrongly ACCEPTS a weak password, accounts are easier to
     guess or brute-force. This is the more serious direction of failure.
     
   - User experience impact: if it wrongly REJECTS a good password, people get blocked at
     sign-up or get confusing errors, and some give up.
     
   - Business impact: weaker accounts raise the chance of breaches, support load and
     reputation damage; rejected sign-ups mean lost customers.


2. Risk level: □ Low  x [Medium]  □ High

3. Based on this risk, my testing strategy should be:
   □ Minimal - basic happy path only
   x [Standard - equivalence classes + boundaries]
   □ Comprehensive - all edge cases, stress tests

"""
````

### Step 2: Equivalence Partitioning

````python
"""
Identify equivalence classes (categories that behave the same):

VALID class (all requirements met):
- Example: "Pass123!"  (8 chars, has uppercase, number and special character)

INVALID classes (each requirement violation):
- Too short (< 8 chars): Example: "Pass12!"  (7 chars, everything else is fine)
- Missing uppercase: Example: "pass123!"
- Missing number: Example: "Password!"
- Missing special char: Example: "Pass1234"
- Multiple violations: Example: "pass"  (short, no uppercase, no number, no special)

Do I need to test EVERY invalid combination?
Decision: No
Reasoning: With 4 independent rules there are 15 ways to be invalid, but each rule is its own
separate check. One test that breaks ONLY that rule proves the check works (and a bug in one
check would show up in its own test). One multiple-violation case shows the checks do not
cancel each other out. Testing all 15 combinations adds almost no extra confidence for a lot
more tests to read and maintain.

"""
````

### Step 3: Boundary Analysis

````python
"""
Key boundaries to test:

Length boundary (8 characters):
- 7 chars (just under): "Pass12!"    -> False
- 8 chars (exactly at): "Pass123!"   -> True
- 9 chars (just over): "Pass1234!"  -> True

Other boundaries?
- Empty string?
  Yes: "" -> False. A very common place for crashes, and the cheapest test there is.
  
- Very long password (100+ chars)?
Yes, one test: "A1!" followed by 100 "a" characters -> True.
  It checks there is no hidden maximum length. No further stress testing at this risk level.

- Only special characters?
  Yes: "!!!!!!!!" -> False (long enough and has a special character,
  but no uppercase and no number).
""" 
````

### Step 4: Test Adequacy Decision

````python
"""
How many tests is enough?

Minimum viable: 6 tests (one per equivalence class: 1 valid + 5 invalid)
My plan: 11 tests

Why this number?
The 6 class tests are the floor. The extra 5 each guard against a different, realistic bug:
the 9-character case documents that "at least 8" means 8 or more; the empty string guards
against crashes; special-characters-only guards against a function that stops checking once
it finds one rule satisfied; "Pass123?" guards against accepting characters outside the
allowed set; and the 100+ character case guards against a hidden maximum length.

Could I get the same confidence with fewer tests?
Slightly. The 7-char and 8-char pair already pins the length boundary down, so the 9-char
test is the most redundant, and I would drop it first. Everything else catches something
that nothing else catches.

What am I NOT testing, and why is that OK?
- All 15 invalid rule combinations: the rules are independent (see above).
- Non-English letters, emoji, spaces and tabs: the requirements do not mention them.
- None or non-string input: the requirements say the input is a password string.
- Common-password lists ("Password1!") or a maximum length: not in the requirements.
- Performance on huge inputs: Medium risk and a single pass over the string, so not worth it.

"""
````

### Step 5: Write Specifications

````python
# Based on your analysis, write focused assertions:

def is_strong_password(password):
    """Validate password strength."""
    allowed_special = "!@#$%^&*"
    return (
        len(password) >= 8
        and any(char.isupper() for char in password)
        and any(char.islower() for char in password)
        and any(char.isdigit() for char in password)
        and any(char in allowed_special for char in password)
    )

# Valid password (meets all requirements):
assert is_strong_password("Pass123!") == True        # exactly 8 chars (boundary: at)

# Invalid passwords (one per major equivalence class):
assert is_strong_password("Pass12!") == False        # too short (boundary: just under)
assert is_strong_password("pass123!") == False       # missing uppercase
assert is_strong_password("Password!") == False      # missing number
assert is_strong_password("Pass1234") == False       # missing special character
assert is_strong_password("pass") == False           # multiple violations

# Boundary tests (if risk justifies):
assert is_strong_password("Pass1234!") == True       # 9 chars (just over)
assert is_strong_password("") == False               # empty string
assert is_strong_password("!!!!!!!!") == False       # only special characters
assert is_strong_password("Pass123?") == False       # "?" is not an allowed special char
assert is_strong_password("A1!" + "a" * 100) == True # 103 chars, no hidden maximum length
````

**Step 2:** Get AI's implementation and paste it below:

````python
# AI's implementation:
def is_strong_password(password):
    """Return True if the password meets all four strength rules."""
    special_characters = "!@#$%^&*"

    # Rule 1: at least 8 characters
    if len(password) < 8:
        return False

    has_uppercase = False
    has_number = False
    has_special = False

    # Check each character once and note which kinds we have seen
    for character in password:
        if character.isupper():
            has_uppercase = True
        if character.isdigit():
            has_number = True
        if character in special_characters:
            has_special = True

    # Rules 2-4: all three kinds of character must be present
    return has_uppercase and has_number and has_special
````

**Step 3:** Test it with your assertions:

````python
# Run your assertions here
password_tests = [
    ("Pass123!", True),          # valid: exactly 8 chars (boundary: at)
    ("Pass1234!", True),         # valid: 9 chars (boundary: just over)
    ("A1!" + "a" * 100, True),   # valid: 103 chars, no hidden maximum length
    ("Pass12!", False),          # too short: 7 chars (boundary: just under)
    ("pass123!", False),         # missing uppercase
    ("Password!", False),        # missing number
    ("Pass1234", False),         # missing special character
    ("pass", False),             # multiple violations
    ("", False),                 # empty string
    ("!!!!!!!!", False),         # only special characters
    ("Pass123?", False),         # "?" is not in the allowed special set
]

all_passed = True
for password, expected in password_tests:
    result = is_strong_password(password)
    if result != expected:
        all_passed = False
        print(f"FAILED: is_strong_password({password!r}) returned {result}, expected {expected}")

print("All tests passed!" if all_passed else "Some tests failed")
````
