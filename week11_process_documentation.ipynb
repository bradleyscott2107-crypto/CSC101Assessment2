# Problem 1: Tip Calculator 

**Step 1: Understand the problem**

This program helps staff at a perth restuarant work out a tip quickly. The staff member types in the bill amount and the tip percentage the customer wants to give, and the program works out how much the tip is and what the total charge should be. 
The program calculates the bill plus the tip.



**Step 2: Inputs and Outputs**

**Inputs:**

- The staff member provides the bill amount (the cost of the meal before any tip) which is $45.80.
- The tip percentage (how much tip the customer wants to leave) which is 15%.
- Both inputs are number. It could be whole numbers or decimals. 

**Outputs:**

- The tip amount.
- The total amount to charge which is the bill plus the tip.
- Bothe outputs are dollar amounts shown to 2 decimal places. 

**Step 3: Work it by Hand**

Bill amount = $45.80, Tip percentage = 15%

1. Turn the percentage into a decimal: 15% = 15÷100 = 0.15
2. Work out the tip by multiplying the bill by that decimal: $45.80 x 0.15 = $6.87
3. Add the tip to the original bill to get the total: $45.80 + $6.87 = $52.67
4. "The tip is $6.87, and the total charge is $52.67"

# Problem 2: Late fee calculator 

**Step 1: Understand the Problem**

This program helps a librarian work out the fine on a book that has been returned late. The librarian types in how many days the book is overdue, and the program multiplies that by $0.50 per day to give the fine. There is a limit though: the fine can never be more than $10.00, so if the amount would be higher than that, the program should say "Maximum fine reached" instead of charging more. 

**Step 2: Identify inputs and outputs**

**Inputs:**

- The librarian provides the number of days the book is overdue. This is a whole number. It counts the days after the 14-day loan period has ended. 

**Outputs:**

- The program calculates the fine amount in dollars, calculated as days overdue x $0.50 and shown to 2 deciaml places. 

**Special cases:**

- Maximum fine: The fine is capped at $10.00. This is reached at 20 days overdue. If the fine would go over $10.00, the program shows the maximum fine and the message "Maximum fine reached".
- Zero days late: if the book is returned on time, the fine is $0.00

**Step 3: Work it by hand**

**Case A: Book is 5 days late**

- My calculation: 5 days x $0.50 per day = $2.50. T
- Fine: $2.50

**Case B: Book is 25 days late**

- My calculation: 25 days x $0.50 per day = $12.50. This is more than the $10.00 limit, so the fine is capped at $10.00
- Fine: $10.00 (Maximum fine reached)
- Special note: This case is different because the normal calculation ($12.50) goes over the maximum. The program has to check the result against the $10.00 limit and use the limit instead, then tell the librarian that the maximum has been reached. 

**Case C: Book is 0 days late**

- My calculation: 0 days x $0.50 per day = $0.00.The book was returned on time, so there is nothing to pay. 
- Fine: $0.00.
