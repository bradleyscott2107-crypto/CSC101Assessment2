## Part B — Nested (Stretch)

Give each contact both **phone** and **email** using a nested dictionary.

### B1. Convert contacts to a nested structure

```python
# TODO: Redesign `contacts` so each entry stores both a phone and an email
contacts = {
    "Brad": {"phone": "0435 999 999", "email": "brad@example.com"},
    "Lewis": {"phone": "0435 771 800", "email": "lewis@example.com"},
    "Peter": {"phone": "0435 771 801", "email": "peter@example.com"},
    "Anthony": {"phone": "0435 771 802", "email": "anthony@example.com"}
}
```

### B2. Print contacts neatly (name, phone, email)

```python
# TODO: Loop through contacts and print both phone and email
for name, details in contacts.items():
    print(f"{name}: {details['phone']} | {details['email']}")
    
```

\::: callout-note
**Optional:** Add a simple **lookup** by name using `input()` to fetch phone/email.
\:::

```python
# Prompt for a name and print their details if found
name = input("Enter a contact name:")
details = contacts.get(name)

if details:
    print(f"Phone: {details['phone']}")
    print(f"Email: {details['email']}")
else:
    print("Contact not found.")
```
