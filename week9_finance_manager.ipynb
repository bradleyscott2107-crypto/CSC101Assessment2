Function 1: Read File

```python
#This code reads the file
def load_transactions(filename):
    """Read lines from a transaction file safely."""
    try:
        with open(filename, "r") as file:
            lines = file.readlines()
        return [line.strip() for line in lines if line.strip()]
    except FileNotFoundError:
        print(f"File '{filename}' not found. Check that it is in the notebook's folder.")
        return []

```

```python
#Test function 1
transaction_lines = load_transactions("transactions.txt.txt")
for line in transaction_lines:
    print(line)
```

How it Works:

-try / except FileNotFoundError – if the file doesn't exist, the program prints a friendly message and returns an empty list instead of crashing.

-with open(filename, 'r') as file – opens the file for reading and automatically closes it afterwards, even if an error happens.

-file.readlines() – reads the file into a list where each item is one line (still with the \n at the end).
 
-[line.strip() for line in lines if line.strip()] – a list comprehension that removes the newline/spaces from each  line (strip()) and drops blank lines (an empty string is “falsy”, so if line.strip() filters it out).

Function 2:Process Data 

```python
def process_transactions(lines):
    """Convert valid transaction lines into dictionaries."""
    transactions = []
    failed_lines = 0

    for line in lines:
        if not line.strip():  # Ignore empty lines
            continue

        try:
            description, amount_text = line.split(",", 1)
            description = description.strip()
            amount = float(amount_text.strip())

            if not description:
                raise ValueError("Missing description")

            transactions.append({
                "description": description,
                "amount": amount
            })

        except ValueError:
            print(f"Skipping invalid line: {line}")
            failed_lines += 1

    print(f"Invalid lines skipped: {failed_lines}")
    return transactions
```

```python
#Test function 2
raw_lines = load_transactions("transactions.txt.txt")
transactions = process_transactions(raw_lines)

for item in transactions:
    print(item)
```

Function 3: Simple Summary 

```python
def show_summary(transactions):
    total_transactions = len(transactions)
    total_income = sum(
        item["amount"] for item in transactions if item["amount"] > 0
    )
    total_expenses = sum(
        item["amount"] for item in transactions if item["amount"] < 0
    )

    print(f"Total transactions: {total_transactions}")
    print(f"Total income: ${total_income:.2f}")
    print(f"Total expenses: ${total_expenses:.2f}")
```

Putting it all together 

```python
def show_summary(transactions):
    total_transactions = len(transactions)
    total_income = sum(
        item["amount"] for item in transactions if item["amount"] > 0
    )
    total_expenses = sum(
        item["amount"] for item in transactions if item["amount"] < 0
    )
    net_amount = total_income + total_expenses

    print("=" * 35)
    print("       TRANSACTION SUMMARY")
    print("=" * 35)
    print(f"Transactions processed: {total_transactions}")
    print(f"Total income:           ${total_income:,.2f}")
    print(f"Total expenses:         ${abs(total_expenses):,.2f}")
    print(f"Net amount:             ${net_amount:,.2f}")
    print("=" * 35)
    
```

```python
show_summary(transactions)
```
