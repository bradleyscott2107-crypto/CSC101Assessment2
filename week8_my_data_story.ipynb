This notebook reads a personal text file containing my favourite movies, displays it, and calculates some statistics about it.

```python
#Step 1: Read the Text File
# Import pathlib so we can check that the file exists before reading it
from pathlib import Path

# The name of the file we want to analyze.
filename = "my_data.txt"
```

```python
#Step 2: Display the contents of the file
# Turn the filename into a Path object so Python can work with file locations.
# ...existing code...
file_path = Path(filename)

# Check whether the file exists before trying to read it.
if file_path.exists():
    # Read the file as text using UTF-8 encoding so it handles normal text correctly.
    movie_text = file_path.read_text(encoding="utf-8")
     # Print the contents of the file to the screen.
    print(movie_text)
else:
   # If the file is missing, print a helpful error message with the full path. 
    print(f"File not found: {file_path.resolve()}")
```

```python
# Step 3: shows 2 pieces of information about the file 
# Import Path again so we can create a file path object.
from pathlib import Path

# Store the name of the file we want to analyse.
filename = "my_data.txt"
file_path = Path(filename)

if file_path.exists():
    text = file_path.read_text(encoding="utf-8")

    number_of_lines = len(text.splitlines())  #1. Count how many lines are in the file
    number_of_words = len(text.split())       #2. Count how many words are in the file

# Print the file name and the statistics.
    print(f"File name: {file_path.name}")
    print(f"Number of lines: {number_of_lines}")
    print(f"Number of words: {number_of_words}")
else:
    print(f"File not found: {file_path.resolve()}")
```
