## Practice Exercise 1.1: Sales Performance by Region

::: {.callout-tip}
## Your AI Collaboration Challenge

**Business Question**: "Which region is performing best, and what's the average sale size by region?"

**Your Task**: Practice prompting AI with this business context. Try creating a prompt that asks for:
1. Total sales by region
2. Average sale amount by region  
3. Number of transactions by region
4. Results formatted for a manager presentation

**Write your prompt below, then see what AI generates:**

**Your Prompt:**
```
I'm a business analyst for an office supply company. My manager wants to know
which region is performing best and what the average sale size is in each one.

I have a pandas DataFrame called sales_df loaded from sales_data.csv with columns
Date, Product, Quantity, Unit_Price, Total_Sale, Sales_Rep, Region. Quantity,
Unit_Price and Total_Sale are stored as text with $ signs, so clean them first.

Write simple, well-commented code that gives me, by region:
1. Total sales
2. Average sale amount
3. Number of transactions

Sort by total sales (highest first), format currency as $X,XXX.XX, and make the
table clean enough to paste into a manager presentation. Then give me a
one-sentence takeaway about the best-performing region.
```

**AI Response:**

````python
# Paste AI-generated code here and run it
# Remember to critique and improve if needed!

import pandas as pd

sales_df = pd.read_csv('sales_data.csv')

# Clean: strip $ and commas, convert to numbers
for col in ['Quantity', 'Unit_Price', 'Total_Sale']:
    sales_df[col] = pd.to_numeric(sales_df[col].str.replace(r'[$,]', '', regex=True))

# Summarise by region
region_summary = (
    sales_df.groupby('Region')
    .agg(Total_Sales=('Total_Sale', 'sum'),
         Average_Sale=('Total_Sale', 'mean'),
         Transactions=('Total_Sale', 'count'))
    .sort_values('Total_Sales', ascending=False)
)

# Format for presentation
report = region_summary.copy()
report['Total_Sales'] = report['Total_Sales'].map('${:,.2f}'.format)
report['Average_Sale'] = report['Average_Sale'].map('${:,.2f}'.format)

print("SALES PERFORMANCE BY REGION")
print("=" * 50)
print(report)

best = region_summary.index[0]
share = region_summary.loc[best, 'Total_Sales'] / region_summary['Total_Sales'].sum()
print(f"\nTakeaway: {best} leads with {share:.1%} of total sales.")
````

```

Quantity, Unit_Price, Total_Sale, Sales_Rep, Region. Quantity, Unit_Price and

Write simple, commented code that shows two top-5 tables:
1. Top 5 products by total revenue (include % of overall revenue)
2. Top 5 products by total units sold

Also check whether I can estimate profit margins from this data. If I can't (there's
no cost column), tell me what's missing and suggest the closest useful proxy
instead of inventing numbers. Format currency as $X,XXX.XX and finish with a
```

**AI Response:**

````python
# AI-generated code goes here
import pandas as pd

sales_df = pd.read_csv('sales_data.csv')
for col in ['Quantity', 'Unit_Price', 'Total_Sale']:
    sales_df[col] = pd.to_numeric(sales_df[col].str.replace(r'[$,]', '', regex=True))

products = sales_df.groupby('Product').agg(
    Revenue=('Total_Sale', 'sum'),
    Units=('Quantity', 'sum'),
    Avg_Unit_Price=('Unit_Price', 'mean'),
    Sales_Count=('Total_Sale', 'count'))

# Top 5 by revenue (with share of total)
top_revenue = products.sort_values('Revenue', ascending=False).head(5).copy()
top_revenue['Share_of_Revenue'] = (top_revenue['Revenue'] / products['Revenue'].sum()).map('{:.1%}'.format)

# Top 5 by units
top_units = products.sort_values('Units', ascending=False).head(5)

print("TOP 5 BY REVENUE"); print(top_revenue[['Revenue', 'Units', 'Share_of_Revenue']])
print("\nTOP 5 BY UNITS");  print(top_units[['Units', 'Revenue']])

# Margin check
print("\nCost column available:", 'Unit_Cost' in sales_df.columns)
print("Max sales per product:", products['Sales_Count'].max())
````

## Practice Exercise 2.1: Spending Trend Analysis

::: {.callout-tip}
## Business Context

You're helping a friend understand their spending habits over time.

**Challenge**: Ask AI to help you:
1. Convert dates to datetime format
2. Analyze spending trends by day of week
3. Identify if spending has increased over time
4. Spot any unusual spending patterns

**Focus**: Practice asking AI to explain any complex date handling code it provides.
:::

**Your Analysis:**

My Prompt: 

I'm helping a friend understand their spending habits. I have a pandas DataFrame
called transactions_df loaded from transactions.csv with columns Date, Amount,
Category, Description. Amount is stored as text with $ signs, and refunds are
negative (e.g. -$25.00). Date is currently a string.

Write simple, commented code that:
1. Converts Date to datetime and cleans Amount into numbers
2. Analyses spending by day of week (total, average and number of purchases),
   keeping refunds separate from spending
3. Checks whether spending has increased over time, using full weeks only and
   telling me if there isn't enough data to say
4. Flags any unusual purchases and explains how you decided what counts as unusual

Please explain each date-handling line in plain English, including what
dt.day_name() and the week grouping do, and why I need to convert to datetime
first. Keep the explanation suitable for a beginner.

````python
# Your AI collaboration for time-based analysis

import pandas as pd

transactions_df = pd.read_csv('transactions.csv')

# Clean Amount: remove $ and commas, then convert text to numbers
transactions_df['Amount'] = pd.to_numeric(
    transactions_df['Amount'].str.replace(r'[$,]', '', regex=True))

# Convert Date from text to real dates, so Python knows 2024-08-05 is a Monday
transactions_df['Date'] = pd.to_datetime(transactions_df['Date'])

# .dt gives access to date parts; day_name() returns "Monday", "Tuesday", etc.
transactions_df['Day'] = transactions_df['Date'].dt.day_name()

# Keep spending (positive) separate from refunds (negative)
spending = transactions_df[transactions_df['Amount'] > 0]

# Spending by day of week, in calendar order
order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
by_day = (spending.groupby('Day')['Amount']
          .agg(Total='sum', Average='mean', Purchases='count')
          .reindex(order))
print(by_day.round(2))


# Confirm the conversion worked (Date should now be datetime64)
print(transactions_df.dtypes)

midpoint = spending['Date'].min() + (
    spending['Date'].max() - spending['Date'].min()
) / 2

# First half vs second half, formatted
first_half = spending[spending['Date'] <= midpoint]['Amount'].sum()
second_half = spending[spending['Date'] > midpoint]['Amount'].sum()
print(f"First half: ${first_half:,.2f} | Second half: ${second_half:,.2f}")

q1 = spending['Amount'].quantile(0.25)
q3 = spending['Amount'].quantile(0.75)

# Outlier check, with a message if nothing is found
upper_fence = q3 + 1.5 * (q3 - q1)
outliers = spending[spending['Amount'] > upper_fence]
if outliers.empty:
    print(f"No purchases above the outlier threshold (${upper_fence:,.2f})")
else:
    print(outliers)

# Second check: compare purchases within a category (coffee)
print(spending[spending['Category'] == 'Coffee'][['Date', 'Amount', 'Description']])
````

## Notes 
Monday had the highest spending ($298), but that's because both big grocery shops fell on Mondays.
Weeks 31 and 34 are partial (4 days and 2 days), so only weeks 32 and 33 can be compared. Spending rose from $325.45 to $406.60 across those two weeks, and the second half is about 5% higher, so spending is roughly flat to slightly up. Three weeks isn't enough to be sure.
Nothing was flagged as an outlier, but the Starbucks coffee stands out within its category.

## Practice Exercise 2.2: Budget Category Analysis

::: {.callout-tip}
## Personal Finance Scenario

Someone wants to set a realistic monthly budget based on their spending history.

**Challenge**: Use AI to help:
1. Calculate average monthly spending by category
2. Identify categories with high variation (inconsistent spending)
3. Suggest realistic budget amounts for each category
4. Find categories where spending might be reduced

**Advanced**: Ask AI to explain the statistical concepts it uses.
:::

**Your Budget Analysis:**

My Prompt: 

Someone wants to set a realistic monthly budget from their spending history. I have a
pandas DataFrame transactions_df from transactions.csv (columns: Date, Amount,
Category, Description). Amount is cleaned to numbers, Date is datetime, and refunds
are negative. Use only positive amounts as spending.

Write simple, commented code that gives, per category:
1. Total spending, number of purchases, average and median purchase
2. How much spending varies (coefficient of variation)
3. An estimated monthly amount and a suggested budget

The data only covers 1-20 August, so tell me whether a true monthly average is
possible. If not, scale up to a 31-day month, treat once-a-month bills (like
utilities) as a single bill instead of scaling them, and flag where the estimate
is unreliable. Then identify which categories are discretionary and where
spending could realistically be reduced.

Explain each statistical concept you use (mean vs median, standard deviation,
coefficient of variation) in one or two plain-English sentences. Show currency
as $X,XXX.XX and present the results as a clean table.

````python
# AI-assisted budget recommendation code

import numpy as np

days = (transactions_df['Date'].max() - transactions_df['Date'].min()).days + 1  # 20 days

cat = (spending.groupby('Category')['Amount']
       .agg(Total='sum', Purchases='count', Average='mean', Median='median', Std='std'))

# Coefficient of variation = std / mean (higher = less consistent spending)
cat['CV'] = cat['Std'] / cat['Mean'] if 'Mean' in cat else cat['Std'] / cat['Average']
cat['Share'] = cat['Total'] / cat['Total'].sum()

# Scale 20 days up to 31, except once-a-month bills
monthly_bills = ['Utilities']
cat['Est_Monthly'] = np.where(cat.index.isin(monthly_bills), cat['Total'], cat['Total'] / days * 31)

# Suggested budget: estimate rounded up to the nearest $10
cat['Suggested_Budget'] = np.ceil(cat['Est_Monthly'] / 10) * 10

print(cat.sort_values('Total', ascending=False).round(2))
print("Suggested total budget:", cat['Suggested_Budget'].sum())
````

````python
# AI-assisted personal finance advice generation

import pandas as pd

# ASSUMED benchmarks (share of tracked spending). Placeholders: replace with a sourced benchmark.
TYPICAL_SHARE = {'Groceries': 0.35, 'Utilities': 0.15, 'Transport': 0.15,
                 'Entertainment': 0.15, 'Dining': 0.15, 'Coffee': 0.05}
ESSENTIAL = {'Groceries', 'Utilities', 'Transport'}
TOLERANCE = 0.02       # allow 2 percentage points above typical before flagging
MIN_PURCHASES = 2      # need at least 2 purchases to comment on a category

def coaching_report(transactions, typical_share=TYPICAL_SHARE):
    """Friendly spending feedback from cleaned transactions."""
    spending = transactions[transactions['Amount'] > 0]       # exclude refunds
    total = spending['Amount'].sum()
    cats = spending.groupby('Category')['Amount'].agg(Total='sum', Purchases='count')
    cats['Share'] = cats['Total'] / total

    advice = []
    for cat, row in cats.sort_values('Total', ascending=False).iterrows():
        typical = typical_share.get(cat)
        n = int(row['Purchases'])
        if typical is None or n < MIN_PURCHASES:
            advice.append(f"{cat}: not enough data yet to give advice "
                          f"({n} purchase{'s' if n != 1 else ''}).")
            continue
        gap = row['Share'] - typical
        if gap > TOLERANCE:
            excess = gap * total                               # $ above the typical level
            if cat in ESSENTIAL:
                advice.append(f"{cat}: {row.Share:.1%} of spending (typical {typical:.0%}). "
                              f"This is an essential, so try planning shops or comparing prices "
                              f"- matching the typical level would be about ${excess:,.2f} less.")
            else:
                advice.append(f"{cat}: {row.Share:.1%} of spending (typical {typical:.0%}). "
                              f"Cutting back to the typical level would free up about ${excess:,.2f}.")
        else:
            advice.append(f"Nice work! {cat} is {row.Share:.1%} of spending, in line with "
                          f"or below the typical {typical:.0%}.")

    days = (spending['Date'].max() - spending['Date'].min()).days + 1
    print(f"YOUR SPENDING FEEDBACK ({days} days, total ${total:,.2f})")
    print("=" * 60)
    for line in advice:
        print("-", line)
    print("-" * 60)
    print("General information only, not financial advice.")

coaching_report(transactions_df)
````
